# Reto 01: Análisis de datos meteorológicos

Soy analista del Ayuntamiento y voy a analizar los datos diarios de una
estación meteorológica durante 2024. El fichero es un CSV con una fila por día.

## 2. Carga de datos

El fichero tiene 365 filas (una por día de 2024) y estos campos:

- **fecha**: día del registro, en formato AAAA-MM-DD.
- **temperatura_max**: temperatura máxima del día, en ºC.
- **temperatura_min**: temperatura mínima del día, en ºC.
- **precipitacion_mm**: lluvia acumulada ese día, en milímetros.

```python
import csv

def cargar_csv_a_lista_dicts(ruta):
    datos = []
    with open(ruta, newline="", encoding="utf-8") as f:
        lector = csv.DictReader(f)
        for fila in lector:
            datos.append(fila)
    return datos

datos = cargar_csv_a_lista_dicts("estacion_meteorologica_2024_sin_media.csv")

# Las 5 primeras filas
for fila in datos[:5]:
    print(fila)
```

```python
errores = 0

for fila in datos:
    # Pasamos de texto a número
    tmax = float(fila["temperatura_max"])
    tmin = float(fila["temperatura_min"])
    lluvia = float(fila["precipitacion_mm"])

    # Si algo se sale de lo permitido, lo contamos y lo mostramos
    if tmax < -50 or tmax > 60 or tmin < -50 or tmin > 60 or lluvia < 0:
        errores += 1
        print("Dato erróneo:", fila)

print("Total de filas:", len(datos))
print("Filas erróneas:", errores)
```

**Comprobación de datos:** he recorrido las 365 filas y no hay ningún dato erróneo:
todas las temperaturas están entre -50 ºC y 60 ºC y no hay precipitaciones negativas.
Por tanto, todos los datos son válidos y puedo usarlos tal cual.

## 3. Procesamiento de datos

```python
suma_medias = 0

for fila in datos:
    tmax = float(fila["temperatura_max"])
    tmin = float(fila["temperatura_min"])
    media_dia = (tmax + tmin) / 2
    suma_medias += media_dia

temp_media = suma_medias / len(datos)
print("Temperatura media del periodo:", round(temp_media, 2), "ºC")
```

```python
temp_max_total = float(datos[0]["temperatura_max"])
temp_min_total = float(datos[0]["temperatura_min"])

for fila in datos:
    tmax = float(fila["temperatura_max"])
    tmin = float(fila["temperatura_min"])

    if tmax > temp_max_total:
        temp_max_total = tmax
    if tmin < temp_min_total:
        temp_min_total = tmin

print("Temperatura máxima registrada:", temp_max_total, "ºC")
print("Temperatura mínima registrada:", temp_min_total, "ºC")
```

```python
precip_total = 0

for fila in datos:
    precip_total += float(fila["precipitacion_mm"])

precip_media = precip_total / len(datos)

print("Precipitación total:", round(precip_total, 1), "mm")
print("Precipitación media diaria:", round(precip_media, 2), "mm")
```

```python
X = 10
dias_lluvia = 0

for fila in datos:
    if float(fila["precipitacion_mm"]) > X:
        dias_lluvia += 1

print("Días con más de", X, "mm:", dias_lluvia)
```

```python
Y = 30
Z = 5
dias_calor = 0
dias_frio = 0

for fila in datos:
    media_dia = (float(fila["temperatura_max"]) + float(fila["temperatura_min"])) / 2

    if media_dia > Y:
        dias_calor += 1
    if media_dia < Z:
        dias_frio += 1

print("Días con media >", Y, "ºC:", dias_calor)
print("Días con media <", Z, "ºC:", dias_frio)
```

## 4. Análisis

**¿Hubo más días calientes que fríos?** (calientes: media > 25 ºC, fríos: media < 10 ºC)

```python
dias_calientes = 0
dias_frios = 0

for fila in datos:
    media_dia = (float(fila["temperatura_max"]) + float(fila["temperatura_min"])) / 2

    if media_dia > 25:
        dias_calientes += 1
    if media_dia < 10:
        dias_frios += 1

print("Días calientes (media > 25 ºC):", dias_calientes)
print("Días fríos (media < 10 ºC):", dias_frios)

if dias_calientes > dias_frios:
    print("Hubo más días calientes que fríos.")
elif dias_frios > dias_calientes:
    print("Hubo más días fríos que calientes.")
else:
    print("Hubo los mismos días calientes que fríos.")
```

**Respuesta:** sí, hubo bastantes más días calientes (100) que fríos (24).
El clima de 2024 fue cálido en general: casi 3 meses y medio con media por encima de 25 ºC.

## 5. Conclusiones y recomendaciones

**Aspectos meteorológicos relevantes para el municipio:**

- **Calor:** la temperatura media del año fue de 20,13 ºC, con una máxima de 38,6 ºC
  y 36 días con media por encima de 30 ºC. Conviene preparar medidas para las olas
  de calor (zonas de sombra, avisos a la población, actividades al aire libre
  en horarios más frescos).
- **Frío:** la mínima fue de 2,6 ºC y no hubo ningún día con media por debajo de 5 ºC,
  así que el frío extremo no es un problema.
- **Lluvia:** cayeron 993,4 mm en total, pero muy mal repartidos: hubo 39 días con
  más de 10 mm (lluvias fuertes), y la mayoría se concentran en otoño e invierno.
  Hay que revisar el alcantarillado y los planes de prevención de inundaciones.
- **Agua y riego:** en primavera y verano llovió muy poco, por lo que se recomienda
  ahorrar agua y guardar la de los meses lluviosos para el riego.

**Limitaciones del estudio:**

- Solo se analiza un año (2024), así que no se puede saber si es un año normal
  o raro.
- Los datos vienen de una sola estación, que puede no representar todo el municipio.
- La temperatura media diaria se ha calculado como (máxima + mínima) / 2,
  que es una aproximación y no la media real de todo el día.
- No hay datos de otras variables importantes, como el viento o la humedad.
