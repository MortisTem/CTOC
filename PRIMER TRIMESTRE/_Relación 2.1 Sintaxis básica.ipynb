```python
#1.1
#Declaro las variables de cada tipo.
entero = 10
decimal = 3.5
cadena = "Leo Messi"
booleano = True

#Se muestran por pantalla.
print(entero)
print(decimal)
print(cadena)
print(booleano)
#-----------------------------------------------------------------------------------------------------#
#1.2
#Se pide al usuario su nombre y edad.
nombre = input("Introduce tu nombre: ")
edad = input("Introduce tu edad: ")

#Muestro el mensaje.
print(f"Hola, señor {nombre}, usted tiene {edad} años.")
#-----------------------------------------------------------------------------------------------------#
#1.3
#Con esto pido un número decimal al usuario.
numero = float(input("Introduce un número decimal: "))

#Se convierte a entero.
numeroEntero = int(numero)

#Muestro el resultado.
print(f"El número {numero} convertido a entero es {numeroEntero}")
#-----------------------------------------------------------------------------------------------------#
#1.4
#Declaro dos variables numéricas.
a = 2
b = 6

#Hago las funciones de operaciones.
suma = a + b
resta = a - b
multiplicacion = a * b
division = a / b

#Muestro el resultado de cada una.
print(f"Suma: {a} + {b} = {suma}")
print(f"Resta: {a} - {b} = {resta}")
print(f"Multiplicación: {a} * {b} = {multiplicacion}")
print(f"División: {a} / {b} = {division}")
```


## 2. Operadores
1. Pide dos números al usuario y muestra:  
- Su suma  
- Su división entera  
- El resto de la división  
- La potencia del primero elevado al segundo  
  


---

```python
#2.1
#Pedimos dos números al usuario.
z = int(input("Introduce un número: "))
k = int(input("Introduce otro número: "))

#Se calculan las operaciones pedidas.
suma = z + k
divisionEntera = z // k
resto = z % k
potencia = z ** k

#Se muestran los resultados.
print(f"Suma: {z} + {k} = {suma}")
print(f"División entera: {z} // {k} = {divisionEntera}")
print(f"Resto de la división: {z} % {k} = {resto}")
print(f"Potencia del primero elevado al segundo: {z} ** {k} = {potencia}")
```

## 3. Condicionales Simples
1. Pide un número y muestra si es positivo.  
2. Pide una contraseña y comprueba si coincide con `"admin"`.  
3. Pide una nota y muestra `"Has aprobado"` si es mayor o igual a 5.  
4. Escribe un programa que compruebe si un número es par o impar.  
5. Pide al usuario su edad e imprima si está en el rango de 18 a 65 años.

---

```python
#3.1
#Pido el número.
numero = float(input("Introduzca un número: "))

#Se comprueba si es positivo.
if numero > 0:
    print(f"El número {numero} es positivo")
#-----------------------------------------------------------------------------------------------------#
#3.2
#Pido un contraseña.
contraseña = input("Introduzca una contraseña: ")

#Compruebo si coincide con admin.
if contraseña == "admin":
    print("Contraseña correcta.")
else:
    print("Contraseña incorrecta.")
#-----------------------------------------------------------------------------------------------------#
#3.3
#Se pide una nota al usuario.
nota = float(input("Introduce tu nota: "))

#Se comprueba si es mayor o igual a 5.
if nota >= 5:
    print("Has aprobado.")
#-----------------------------------------------------------------------------------------------------#
#3.4
#Se pide un número al usuario.
numero = int(input("Introduce un número: "))

#Con esto compruebo si es par o impar usando el resto de dividir entre 2.
if numero % 2 == 0:
    print(f"El número {numero} es par.")
else:
    print(f"El número {numero} es impar.")
#-----------------------------------------------------------------------------------------------------#
#3.5
#Pido la edad.
edad = int(input("Introduce tu edad: "))

#Compruebo si está en el rango de 18 a 65.
if edad >= 18 and edad <= 65:
    print(f"Tienes {edad} años, estás en el rango de 18 a 65.")
else:
    print(f"Tienes {edad} años, estás fuera del rango de 18 a 65.")
```


## 4. Condicionales Compuestos (`if / else`)
1. Pide un número y muestra si es positivo, negativo o 0.  
2. Pide la edad de una persona y muestra:  
- `"Eres mayor de edad"` si es mayor o igual a 18.  
- `"Eres menor de edad"` en caso contrario.  


---

```python
#4.1
#Pido introducir un número.
z = int(input("Introduce un número: "))

#Se comprueba si es positivo, negativo o cero.
if z > 0:
    print("Número positivo.")
elif z < 0:
    print("Número negativo.")
else:
    print("Nada, 0.")
#-----------------------------------------------------------------------------------------------------#
#4.2
#Se pide la edad al usuario.
edad = int(input("Introduzca su edad: "))

#Se comprueba si es mayor o menor de edad.
if edad >= 18:
    print("Eres mayor de edad.")
else:
    print("Eres menor de edad.")
```

## 5. Condicionales Múltiples (`if / elif / else`)
1. Pide una nota y clasifícala en:  
- Suspenso (<5)  
- Aprobado (5-6)  
- Notable (7-8)  
- Sobresaliente (9-10)  
2. Pide al usuario un número del 1 al 7 y muestra el día de la semana correspondiente.  
3. Pide una temperatura y muestra:  
- `"Frío"` si es menor de 15.  
- `"Templado"` si está entre 15 y 25.  
- `"Caluroso"` si es mayor de 25.  

---

```python
#5.1
#Pido la nota al usuario.
nota = float(input("Introduce tu nota: "))

#Clasifico la nota.
if nota < 5:
    print("Suspenso.")
elif nota <= 6:
    print("Aprobado.")
elif nota <= 8:
    print("Notable.")
else:
    print("Sobresaliente.")
#-----------------------------------------------------------------------------------------------------#
#5.2
#Pido el número al usuario.
dia = int(input("Introduce un número del 1 al 7: "))

#Muestro el día correspondiente según el número que introduzca el usuario.
if dia == 1:
    print("Lunes.")
elif dia == 2:
    print("Martes.")
elif dia == 3:
    print("Miércoles.")
elif dia == 4:
    print("Jueves.")
elif dia == 5:
    print("Viernes.")
elif dia == 6:
    print("Sábado.")
elif dia == 7:
    print("Domingo.")
else:
    print("Número no válido.")
#-----------------------------------------------------------------------------------------------------#
#5.3
#Pido la temperatura al usuario.
temperatura = float(input("Introduce la temperatura: "))

#Clasifico la temperatura.
if temperatura < 15:
    print("Frío.")
elif temperatura <= 25:
    print("Templado.")
else:
    print("Caluroso.")
```

1. Pide una nota y además comprueba:  
- Si la nota es mayor o igual a 5, pregunta si es mayor o igual a 9 → `"Sobresaliente"`.  
- Si no, simplemente indica `"Aprobado"`.  
2. Pide la edad y si es mayor de 18:  
- Pregunta si tiene carnet de conducir.  
- Si lo tiene → `"Puedes conducir"`.  
- Si no → `"Necesitas carnet para conducir"`.  
3. Pide un usuario y una contraseña:  
- Si el usuario es `"admin"`, comprueba si la contraseña es `"1234"`.  
- Si ambas coinciden → `"Acceso concedido"`.  
- Si no → `"Acceso denegado"`.  


```python
#6.1
#Pido la nota al usuario.
nota = float(input("Introduce tu nota: "))

if nota >= 5:
    if nota >= 9:
        print("Sobresaliente.")
    else:
        print("Aprobado.")
#-----------------------------------------------------------------------------------------------------#
#6.2
#Pido la edad al usuario.
edad = int(input("Introduce tu edad: "))

#Se comprueba si es mayor de 18.
if edad > 18:
    #Se pregunta por el carnet.
    carnet = input("¿Tienes carnet de conducir? (si o no): ")
    
    if carnet == "si":
        print("Puedes conducir.")
    else:
        print("Necesitas carnet para conducir.")
#-----------------------------------------------------------------------------------------------------#
#6.3
#Se pide el usuario y la contraseña.
usuario = input("Introduce tu usuario: ")
contraseña = input("Introduce tu contraseña: ")

#Se comprueba si el usuario es admin y dentro de comprueba la contraseña.
if usuario == "admin":
    if contraseña == "1234":
        print("Acceso concedido.")
    else:
        print("Acceso denegado.")
else:
    print("Acceso denegado.")
```
